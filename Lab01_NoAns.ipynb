### Problem 8
#### Write a function that returns the character that is n positions after character x in the alphabet.
#### Preserve case and wrap around within A..Z or a..z.

```python
x=input()
n=int(input())
def new_character(x,n):
  if x.isupper:
    start=ord('A')
  else:
    start=ord('a')
position=ord(x)-start
new_position=((position+n)%26)
return chr(start+new_position)
print new_character(x,n)
```
