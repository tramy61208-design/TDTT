Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểmtra xem nó có thể làđộ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra mànhình.


```python
a,b,c=map(float,input().split())
if a+b<=c or a+c<=b or b+c<=a:
  print("Khong la ba canh cua mot tam giac")
else:
  if a==b==c:
    print("Tam giac deu")
  elif a==b or a==c or b==c:
    print("Tam giac can")
  elif (a==b or a==c or b==c) and (a*a+b*b==c*c or a*a==b*b+c*c or b*b==c*c+a*a):
    print('Tam giac vuong can')
  elif a*a+b*b==c*c or a*a==b*b+c*c or b*b==c*c+a*a:
    print('Tam giac vuong')
  else:
    print('Tam giac thuong')
  cv=a+b+c
  p=cv/2
  s=(p*(p-a)*(p-b)*(p-c))**0.5
  print(f"{cv:.2f} {s:.2f}")
```

```python
#tam duong tron I(a,b)
a,b=map(float,input().split())
r=float(input())
#diem A(x,y)
x,y=map(float,input().split())
#kiem tra
if (a-x)**2+(b-y)**2==r**2:
  print('A thuoc duong tron')
else:
  print('A khong thuoc duong ')
```

Bài 11: Tính x**y (Với x, y là số thực).

```python
x, y = map(float, input().split())

if x == 0 and y < 0:
    print("Không xác định (lỗi chia cho 0)")
elif x < 0 and not y.is_integer():
    print("Không có kết quả trên tập số thực")
else:
    try:
        print(x ** y)
    except OverflowError:
        print("Kết quả quá lớn (tràn số)")
```
